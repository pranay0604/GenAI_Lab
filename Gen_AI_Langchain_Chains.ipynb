### Structured Prompt

```python
from langchain.prompts import PromptTemplate

structured_prompt = PromptTemplate(
    input_variables = ['topic'],
    template = 'Provide a brief explanation of {topic} and list its three main concepts in a single line'    
    )
```

```python
structured_prompt
```

### Run the chain : LCEL

- Lang Chain Expression Language (LCEL)

- A new way to connect pipelines in langchain framework

- We connect structure prompt and llm using **pipe operator ( | )**

```python
chain = structured_prompt | llm
inputs = {'topic' : 'Hypothesis Testing'}
response = chain.invoke(inputs)
print(response.content)
```

**Old Method**

```python
from langchain.chains import LLMChain

LLMChain(llm=llm, prompt=structured_prompt).run("Hypothesis Testing")

# Here we got 2 warnings

# 1) Warning 1 : related LCEL

# 2) Warning 2 : instead of run use invoke
```

```python
from langchain.chains import LLMChain

LLMChain(llm=llm, prompt=structured_prompt).run("Gemini Model")
```

```python
from langchain.chains import LLMChain

LLMChain(llm=llm, prompt=structured_prompt).invoke("Hypothesis Testing")
```

```python
chain = structured_prompt | llm
print(chain.invoke('transformers').content)
```

### Pipeline Creation

```python
chain = structured_prompt | llm | (lambda x : x.content)
print(chain.invoke("Z Test"))
```

### Parser

- Provide the clean output

- it is available in langchain_core

- the package name is langchain_core
    
    - class : Out_Parsers
    
    - Method Name : StrOutputParser

```python
from langchain_core.output_parsers import StrOutputParser

parser = StrOutputParser()
chain = structured_prompt | llm | parser
print(chain.invoke("Linear Regression"))
```
