- Saving the conversations in the memory

```python
from langchain.memory import ConversationBufferMemory
from langchain.chains import ConversationChain
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv("/Users/pranay/Documents/VS_Projects/streamlit_examples/.env")

import os
groq_api_key = os.getenv("GROQ_API_KEY")

llm = ChatGroq(api_key = groq_api_key,
               model = 'llama-3.3-70b-versatile',
               temperature = 0.7,
               max_tokens = 512
        )
```

```python
memory = ConversationBufferMemory()
```

```python
chain = ConversationChain(llm=llm, 
                          memory=memory, 
                          verbose=True)
```

```python
chain.invoke("My name is Pranay!")
chain.invoke("I lives in Maharashtra")
chain.invoke("I am a Data Scientist student")
```

```python
chain.invoke('Where am I living?')
```

ConversationBufferWindowMemory
----

- The name window suggest it remembers only k turns

- One Turn (1 Human + 1 AI)

- For Example

    - 1. Human : Hello My name is Pranay

    - 2. AI : Hi Pranay

    - 3. Human : I lives in Maharashtra

    - 4. AI : Maharashtra is good

    - 5. AI: What is My name

- suppose the converstaion chain 3 turns

    - turn 1 (1 + 2)

    - turn 2 (3 + 4)

    - turn 3 (5 + AI answer)

- we provided window k=2, only last two converstaions remember

    - mean turn 2 and turn 3 will remember 

    - turn 1 will drop

- old conversation will forgot

```python
from langchain.memory import ConversationBufferMemory
from langchain.memory import ConversationBufferWindowMemory
from langchain.chains import ConversationChain
from langchain_groq import ChatGroq
from dotenv import load_dotenv
load_dotenv("/Users/pranay/Documents/VS_Projects/streamlit_examples/.env")

import os
groq_api_key = os.getenv("GROQ_API_KEY")

llm = ChatGroq(api_key = groq_api_key,
               model = 'llama-3.3-70b-versatile',
               temperature = 0.7,
               max_tokens = 512
        )
```

```python
memory1 = ConversationBufferWindowMemory()
```

```python
chain1 = ConversationChain(llm=llm, 
                          memory=memory1, 
                          verbose=True)
```

```python
chain1.invoke("My name is Pranay!")
chain1.invoke("I lives in Maharashtra")
# chain.invoke("I am a Data Scientist student")
```

```python
chain1.invoke("What is my name?")
```
