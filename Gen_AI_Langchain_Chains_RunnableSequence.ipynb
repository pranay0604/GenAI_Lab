### Structured Prompt

```python
from langchain.prompts import PromptTemplate

structured_prompt = PromptTemplate(
    input_variables = ['topic'],
    template = 'Provide a brief explanation of {topic} and list its three main concepts in a single line'    
    )
```

```python
structured_prompt
```

### Run the chain : LCEL

- Lang Chain Expression Language (LCEL)

- A new way to connect pipelines in langchain framework

- We connect structure prompt and llm using **pipe operator ( | )**

```python
chain = structured_prompt | llm
inputs = {'topic' : 'Hypothesis Testing'}
response = chain.invoke(inputs)
print(response.content)
```

### Parser

- Provide the clean output

### Runnable Sequence

- I have two operations

    - Assume that based on my product I need to generate company name

        - structured_prompt | llm | (lambda x : x.content)
        
        - structured_prompt: Suggest a company name based on my healthcare product

        - llm will take prompt and generate output

        - .content will extract only specific output

    - I want to generate a tag line based on my company name

        - structured_prompt | llm | (lambda x : x.content)

        - structured_prompt: Suggest a tag line name based on my company name
     
        - but company name  coming from above output
     
- This is going in a sequence  

```python
from langchain.schema.runnable import RunnableSequence

name_prompt = PromptTemplate(
    input_variables = ["product"],
    template = 'Provide a company name based on my {product}.'
    )

seq1 = name_prompt | llm | (lambda x : x.content)

tag_line_prompt = PromptTemplate(
    input_variables = ["company name"],
    template = 'Provide a tag line based on my {company name}.'
    )

seq2 = tag_line_prompt | llm | (lambda x : x.content)

Sequence = RunnableSequence(seq1,seq2)

print(Sequence.invoke("Health care invoice details application"))
```

```python
from langchain.schema.runnable import RunnableParallel

name_prompt = PromptTemplate(
    input_variables = ["product"],
    template = 'Provide a company name based on my {product}.'
    )

tag_line_prompt = PromptTemplate(
    input_variables = ["company name"],
    template = 'Provide a tag line based on my {company name}.'
    )

seq1 = name_prompt | llm | (lambda x : x.content)
seq2 = tag_line_prompt | llm | (lambda x : x.content)

dict1 = {'Company Name' : seq1, "Tag Name" : seq2}

parallel = RunnableParallel(dict1)

print(parallel.invoke("Health care invoice details application"))
```
