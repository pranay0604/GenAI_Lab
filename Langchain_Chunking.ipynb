## Chunking :-

- Different LLM models have different token limit, we can not process millions of tokens at a time

- when we load a large text, will split it into smaller pieces is called as Chunks

- These chunks are then Embedded (numerical representation) and stored in a vector database

- So that relevant chunks can be retrived

- Types of Chunking methods :-
    
    - **CharacterTextSplitter**
    
    - **RecursiveCharacterTextSplitter**
    
    - **TokenTextSplitter**
    
    - **SemanticChunking**
    
    - NLTKTextSplitter
    
    - MarkDownTextSplitter
    
    - SpacyTextSplitter
    
    - PythonTextSplitter
    
    - HTMLTextSplitter

### Recursive Text Splitter

- Recursive means hiercachy means step by step process

- It first tries split by paragraphs\n\n).

- If the chunk is still too long it splits by sentence(.)

- If still long then by space or charcter

- Ensure chunks are within the given size but keeps context (using overlap)

- Paragraph ====> Sentence ====> words

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

text = """LangChain is an open-source framework designed to facilitate the development of applications that integrate large language models (LLMs) with external data sources and computational logic.
It provides tools, abstractions, and components to help developers build language model-powered applications beyond simple text generation.
LangChain focuses on enabling chains of operations, where outputs from an LLM can trigger additional logic, database queries, API calls, or document retrieval steps.
This chaining approach allows LLMs to perform multi-step reasoning, question-answering, summarization, and decision-making over complex workflows.
"""

Splitter = CharacterTextSplitter(separator='', 
                                chunk_size = 50, 
                                chunk_overlap = 2)

chunks = Splitter.split_text(text)
chunks
```

```python
from langchain.text_splitter import RecursiveCharacterTextSplitter

separators = ['\n\n','\n',' ','']

text="""Paragraph one: This is the first paragraph. It has two sentences.

Paragraph two: This is the second paragrah which is quite long and will exceed our chunk size limit
for demenostration purpose. It has several sentences to ensure splitting. Here another sentence to make it longer.

Paragraph three: Short."""

splitter = RecursiveCharacterTextSplitter(separators = separators,
                                 chunk_size = 50,
                                 chunk_overlap = 0                                 
                                )

chunks = splitter.split_text(text)
chunks
```

```python
for i,c in enumerate(chunks):
    print(f"Chunk {i} has {len(c)} characters")
```

### TokenTextSplitter

- If we want to chunks based on **tokens** then we use **TokenTextSplitter**

- 'Hello' is one token, but 'Hello' has 5 characters

- CharacterTextSplitter and RecursiveCharacterText splitter based on characters only

- But TokenTextSplitter completely based on token only

    - tokens can be words, special chars

- But directly langchain has method

- **TokenTextSplitter**

- no need to use **tiktoken**

- **tiktoken** will give understanding of how words ==== tokens ==== token ids

- TokenTextSplitter by default will take **gpt-3.5-turbo** i.e **cl100k_base**

```python
from langchain.text_splitter import TokenTextSplitter

text="""Paragraph one: This is the first paragraph. It has two sentences.
Paragraph two: This is the second paragraph which is quite long and will exceed our chunk size limit
for demenostration purpose. It has several sentences to ensure splitting. Here another sentence to make it longer.
Paragraph three: Short."""

splitter = TokenTextSplitter(chunk_size=15,
                            chunk_overlap=3)

# each chunk will have 15 tokens
# each token overlap of 3 tokens

chunks = splitter.split_text(text)
len(chunks)
```

```python
from langchain.text_splitter import TokenTextSplitter

text="""Paragraph one: This is the first paragraph. It has two sentences.
Paragraph two: This is the second paragraph which is quite long and will exceed our chunk size limit
for demenostration purpose. It has several sentences to ensure splitting. Here another sentence to make it longer.
Paragraph three: Short."""

splitter = TokenTextSplitter.from_tiktoken_encoder(model_name='gpt-4',
                                                   chunk_size=15,
                                                   chunk_overlap=3)

# each chunk will have 15 tokens
# each token overlap of 3 tokens

chunks = splitter.split_text(text)
len(chunks)
```
